# Missing Data check
The dataset contains 1000 entries across 9 columns with 3 different data types: int, float, and object. 4 of these columns- Weather, Traffic_level, Time_of_Day, and Courier_Experience_yrs- each have 30 missing values, equating to 120 missing values in all. Of the 1000 entries, 117 rows(about 11.7% of the dataset) have at least one missing entry.
Decision: No imputation for Tasks 1 to 4.
Reason: These tasks use Delivery_Time_min, which has no missing values.
Exception: Task 5 uses Traffic_Level, so it is based on the 970 rows where that column is not missing.
